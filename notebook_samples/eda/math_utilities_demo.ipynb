## `matrix_to_latex`

- **Qué hace**: Convierte un arreglo compatible con NumPy (`ArrayLike`) en una matriz en formato LaTeX dentro del entorno `\begin{pmatrix} ... \end{pmatrix}`. Los arreglos unidimensionales (vectores 1D) son convertidos automáticamente en vectores columna. Los números se redondean al número de decimales indicado y se eliminan ceros finales no significativos (los enteros se representan sin parte decimal).
- **Parámetros**:
  - `A` (`ArrayLike`): Matriz bidimensional o vector unidimensional a convertir.
  - `decimals` (`int`, por defecto `2`): Número máximo de decimales al redondear los valores numéricos.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la matriz utilizando el entorno `pmatrix`.

```python
# 1. Matriz 2x2 con decimales y números enteros
A = np.array([[1.0, 2.5], [3.14159265, 4.0]])
latex_A = matrix_to_latex(A)
print("Matriz 2x2 (código LaTeX):", latex_A)
display(Math(latex_A))

# 2. Vector unidimensional (convertido automáticamente a vector columna)
v = np.array([1, 2, 3])
latex_v = matrix_to_latex(v)
print("Vector columna 1D (código LaTeX):", latex_v)
display(Math(latex_v))

# 3. Matriz de covarianzas 3x3 con 3 decimales
cov_matrix = np.array([
    [1.000, 0.4567, -0.1234],
    [0.4567, 1.000, 0.7891],
    [-0.1234, 0.7891, 1.000]
])
latex_cov = matrix_to_latex(cov_matrix, decimals=3)
print("Matriz 3x3 con 3 decimales:", latex_cov)
display(Math(latex_cov))
```

## `frac_to_latex`

- **Qué hace**: Convierte un numerador y un denominador en una fracción formateada en LaTeX utilizando la macro `\frac{numerador}{denominador}`. Admite tanto valores numéricos (`int`, `float`) como expresiones simbólicas (`str`).
- **Parámetros**:
  - `numerator` (`str | int | float`): Numerador de la fracción.
  - `denominator` (`str | int | float`): Denominador de la fracción.
- **Qué devuelve**:
  - `str`: Expresión LaTeX que representa la fracción.

```python
# 1. Fracción numérica simple
latex_frac_num = frac_to_latex(3, 4)
print("Fracción numérica:", latex_frac_num)
display(Math(latex_frac_num))

# 2. Fracción con expresiones algebraicas
latex_frac_alg = frac_to_latex("x + 1", "x - 2")
print("Fracción algebraica:", latex_frac_alg)
display(Math(latex_frac_alg))

# 3. Fracción con símbolos matemáticos (derivada parcial)
latex_deriv = frac_to_latex(r"\partial \mathcal{L}", r"\partial w_i")
print("Derivada parcial:", latex_deriv)
display(Math(latex_deriv))
```

## `sqrt_to_latex`

- **Qué hace**: Convierte una expresión o valor numérico en una raíz cuadrada en sintaxis LaTeX utilizando `\sqrt{x}`.
- **Parámetros**:
  - `x` (`str | int | float`): Expresión o valor que va dentro del radicando.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la raíz cuadrada.

```python
# 1. Raíz cuadrada de un valor numérico
latex_sqrt_num = sqrt_to_latex(2)
print("Raíz de un escalar:", latex_sqrt_num)
display(Math(latex_sqrt_num))

# 2. Raíz cuadrada de una expresión algebraica
latex_sqrt_alg = sqrt_to_latex("x^2 + y^2")
print("Raíz de expresión algebraica:", latex_sqrt_alg)
display(Math(latex_sqrt_alg))

# 3. Raíz del discriminante cuadrático
latex_sqrt_disc = sqrt_to_latex("b^2 - 4ac")
print("Discriminante:", latex_sqrt_disc)
display(Math(latex_sqrt_disc))
```

## `root_to_latex`

- **Qué hace**: Convierte una expresión o valor y un índice en una raíz n-ésima en sintaxis LaTeX utilizando la macro `\sqrt[index]{x}`.
- **Parámetros**:
  - `x` (`str | int | float`): Expresión o valor dentro del radicando.
  - `index` (`int`): Índice u orden de la raíz.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la raíz n-ésima.

```python
# 1. Raíz cúbica numérica
latex_root_num = root_to_latex(27, 3)
print("Raíz cúbica numérica:", latex_root_num)
display(Math(latex_root_num))

# 2. Raíz cúbica de una expresión algebraica
latex_root_alg = root_to_latex("x + 1", 3)
print("Raíz cúbica algebraica:", latex_root_alg)
display(Math(latex_root_alg))

# 3. Raíz de orden superior (media geométrica para n elementos)
latex_root_n = root_to_latex(r"\prod_{i=1}^n x_i", "n")
print("Raíz n-ésima (media geométrica):", latex_root_n)
display(Math(latex_root_n))
```

## `power_to_latex`

- **Qué hace**: Convierte una base y un exponente en una potencia formateada en sintaxis LaTeX utilizando `base^{exponente}`. El exponente queda correctamente envuelto entre llaves.
- **Parámetros**:
  - `base` (`str | int | float`): Base de la potencia.
  - `exponent` (`str | int | float`): Exponente al que se eleva la base.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la potencia.

```python
# 1. Potencia algebraica básica
latex_pow_basic = power_to_latex("x", 2)
print("Potencia básica:", latex_pow_basic)
display(Math(latex_pow_basic))

# 2. Exponencial con exponente negativo
latex_pow_exp = power_to_latex("e", r"-\lambda t")
print("Exponencial negativa:", latex_pow_exp)
display(Math(latex_pow_exp))

# 3. Potencia con términos compuestos
latex_pow_comp = power_to_latex(r"\sigma", 2)
print("Varianza (sigma al cuadrado):", latex_pow_comp)
display(Math(latex_pow_comp))
```

## `subscript_to_latex`

- **Qué hace**: Añade un subíndice a una expresión base en sintaxis LaTeX utilizando `expresion_{subindice}`.
- **Parámetros**:
  - `expression` (`str`): Expresión o variable base.
  - `subscript` (`str | int`): Valor o símbolo del subíndice.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la expresión con subíndice.

```python
# 1. Variable indexada numéricamente
latex_sub_num = subscript_to_latex("x", 1)
print("Variable indexada:", latex_sub_num)
display(Math(latex_sub_num))

# 2. Pesos o coeficientes matriciales
latex_sub_sym = subscript_to_latex("W", "ij")
print("Elemento de matriz de pesos:", latex_sub_sym)
display(Math(latex_sub_sym))

# 3. Notación temporal en series de tiempo
latex_sub_time = subscript_to_latex("y", "t-1")
print("Lag temporal:", latex_sub_time)
display(Math(latex_sub_time))
```

## `abs_to_latex`

- **Qué hace**: Convierte una expresión o valor numérico en la notación de valor absoluto en LaTeX utilizando delimitadores escalables `\left| x \right|` para ajustarse automáticamente a la altura del contenido.
- **Parámetros**:
  - `x` (`str | int | float`): Expresión o valor cuyo valor absoluto se representa.
- **Qué devuelve**:
  - `str`: Representación LaTeX del valor absoluto con delimitadores escalables.

```python
# 1. Valor absoluto de un número
latex_abs_num = abs_to_latex(-42)
print("Valor absoluto de escalar:", latex_abs_num)
display(Math(latex_abs_num))

# 2. Expresión algebraica de desviación
latex_abs_alg = abs_to_latex(r"x - \bar{x}")
print("Desviación absoluta:", latex_abs_alg)
display(Math(latex_abs_alg))

# 3. Expresión de error absoluto (MAE)
latex_abs_err = abs_to_latex(r"y_i - \hat{y}_i")
print("Error absoluto individual:", latex_abs_err)
display(Math(latex_abs_err))
```

## `norm_to_latex`

- **Qué hace**: Convierte una expresión en la notación de norma vectorial o matricial en LaTeX utilizando delimitadores escalables de doble barra `\left\| x \right\|`.
- **Parámetros**:
  - `x` (`str`): Expresión, vector o matriz cuya norma se representa.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la norma con delimitadores escalables.

```python
# 1. Norma de un vector
latex_norm_vec = norm_to_latex("v")
print("Norma de vector v:", latex_norm_vec)
display(Math(latex_norm_vec))

# 2. Norma de un vector residual en regresión
latex_norm_res = norm_to_latex(r"\mathbf{y} - \mathbf{X}\mathbf{w}")
print("Norma del residuo:", latex_norm_res)
display(Math(latex_norm_res))

# 3. Norma L2 con subíndice (combinada con subscript_to_latex)
latex_norm_l2 = subscript_to_latex(norm_to_latex(r"\mathbf{w}"), 2)
print("Norma L2:", latex_norm_l2)
display(Math(latex_norm_l2))
```

## `transpose_to_latex`

- **Qué hace**: Agrega la notación de traspuesta a una matriz o vector en LaTeX utilizando el superíndice `^{T}` (`A^{T}`).
- **Parámetros**:
  - `A` (`str`): Matriz o expresión vectorial.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la traspuesta.

```python
# 1. Traspuesta de una matriz simple
latex_trans_A = transpose_to_latex("A")
print("Matriz traspuesta:", latex_trans_A)
display(Math(latex_trans_A))

# 2. Traspuesta de una matriz de diseño de características
latex_trans_X = transpose_to_latex(r"\mathbf{X}")
print("Matriz de diseño traspuesta:", latex_trans_X)
display(Math(latex_trans_X))

# 3. Traspuesta de un producto matricial
latex_trans_prod = transpose_to_latex(r"(\mathbf{A}\mathbf{B})")
print("Traspuesta de producto:", latex_trans_prod)
display(Math(latex_trans_prod))
```

## `inverse_to_latex`

- **Qué hace**: Agrega la notación de matriz inversa en LaTeX utilizando el superíndice `^{-1}` (`A^{-1}`).
- **Parámetros**:
  - `A` (`str`): Matriz o expresión matricial.
- **Qué devuelve**:
  - `str`: Representación LaTeX de la matriz inversa.

```python
# 1. Inversa de una matriz simple
latex_inv_A = inverse_to_latex("A")
print("Matriz inversa:", latex_inv_A)
display(Math(latex_inv_A))

# 2. Matriz de covarianza inversa (matriz de precisión)
latex_inv_cov = inverse_to_latex(r"\mathbf{\Sigma}")
print("Matriz de precisión (covarianza inversa):", latex_inv_cov)
display(Math(latex_inv_cov))

# 3. Inversa de matriz Gramiana
latex_inv_gram = inverse_to_latex(r"(\mathbf{X}^{T}\mathbf{X})")
print("Inversa Gramiana:", latex_inv_gram)
display(Math(latex_inv_gram))
```

## `determinant_to_latex`

- **Qué hace**: Convierte una matriz o expresión en la notación funcional de determinante en LaTeX utilizando `\det\left(A\right)`.
- **Parámetros**:
  - `A` (`str`): Matriz o expresión matricial.
- **Qué devuelve**:
  - `str`: Representación LaTeX del determinante.

```python
# 1. Determinante de matriz simbólica
latex_det_A = determinant_to_latex("A")
print("Determinante de A:", latex_det_A)
display(Math(latex_det_A))

# 2. Determinante de la matriz de covarianza
latex_det_cov = determinant_to_latex(r"\mathbf{\Sigma}")
print("Determinante de matriz de covarianza:", latex_det_cov)
display(Math(latex_det_cov))

# 3. Determinante aplicado directamente a una matriz numérica generada
mat_num = np.array([[4, 2], [1, 3]])
latex_det_matrix = determinant_to_latex(matrix_to_latex(mat_num))
print("Determinante de matriz numérica:", latex_det_matrix)
display(Math(latex_det_matrix))
```

## Composición de Fórmulas Matemáticas y Casos de Uso en Data Science

Las funciones de `ds_utils.eda.math_utilities` actúan como componentes modulares (*building blocks*) que pueden combinarse para generar dinámicamente ecuaciones complejas utilizadas en Machine Learning, estadística y optimización.

A continuación se muestran ejemplos prácticos de composición de expresiones analíticas estándar.

```python
# =============================================================================
# 1. Ecuación Normal de Mínimos Cuadrados Ordinarios (OLS)
#    beta_hat = (X^T * X)^{-1} * X^T * y
# =============================================================================
xt = transpose_to_latex(r"\mathbf{X}")
xt_x = f"{xt} " + r"\mathbf{X}"
inv_xt_x = inverse_to_latex(f"({xt_x})")
beta_ols = r"\hat{\boldsymbol{\beta}} = " + f"{inv_xt_x} {xt} " + r"\mathbf{y}"

print("Ecuación Normal OLS:")
print(beta_ols)
display(Math(beta_ols))

# =============================================================================
# 2. Fórmula General de la Ecuación Cuadrática
#    x = (-b \pm \sqrt{b^2 - 4ac}) / (2a)
# =============================================================================
discriminant = sqrt_to_latex(f"{power_to_latex('b', 2)} - 4ac")
num_quadratic = r"-b \pm " + discriminant
den_quadratic = "2a"
formula_cuadratica = f"x = {frac_to_latex(num_quadratic, den_quadratic)}"

print("\nFórmula Cuadrática:")
print(formula_cuadratica)
display(Math(formula_cuadratica))

# =============================================================================
# 3. Distancia de Mahalanobis
#    D_M(x) = \sqrt{(x - \mu)^T * \Sigma^{-1} * (x - \mu)}
# =============================================================================
diff = r"(\mathbf{x} - \boldsymbol{\mu})"
diff_t = transpose_to_latex(diff)
sigma_inv = inverse_to_latex(r"\boldsymbol{\Sigma}")
mahalanobis_interior = f"{diff_t} {sigma_inv} {diff}"
mahalanobis_dist = r"D_M(\mathbf{x}) = " + sqrt_to_latex(mahalanobis_interior)

print("\nDistancia de Mahalanobis:")
print(mahalanobis_dist)
display(Math(mahalanobis_dist))

# =============================================================================
# 4. Función de Pérdida en Regresión Ridge (Error Cuadrático + Regularización L2)
#    J(w) = 1/(2n) * ||y - Xw||^2 + (lambda / 2) * ||w||^2
# =============================================================================
term_mse = frac_to_latex(1, "2n") + " " + power_to_latex(norm_to_latex(r"\mathbf{y} - \mathbf{X}\mathbf{w}"), 2)
term_reg = frac_to_latex(r"\lambda", 2) + " " + power_to_latex(norm_to_latex(r"\mathbf{w}"), 2)
loss_ridge = r"J(\mathbf{w}) = " + f"{term_mse} + {term_reg}"

print("\nFunción de Pérdida Ridge:")
print(loss_ridge)
display(Math(loss_ridge))
```

## Integración con `print_math_section`

La función `print_math_section` de `ds_utils.eda.preview_utilities` permite presentar las expresiones matemáticas generadas junto con un encabezado estilizado y una nota explicativa al pie en formato Markdown.

```python
# Ejemplo de presentación completa combinando math_utilities con preview_utilities
xt_str = transpose_to_latex(r"\mathbf{X}")
inv_gram_str = inverse_to_latex(f"({xt_str}" + r"\mathbf{X})")
ecuacion_presentacion = r"\hat{\boldsymbol{\beta}} = " + f"{inv_gram_str} {xt_str} " + r"\mathbf{y}"

print_math_section(
    content=ecuacion_presentacion,
    header="**Estimador de Mínimos Cuadrados Ordinarios (MCO)**",
    header_separator=True,
    footer="Solución analítica exacta para el vector de coeficientes óptimos que minimiza la suma de cuadrados de los residuos."
)
```
